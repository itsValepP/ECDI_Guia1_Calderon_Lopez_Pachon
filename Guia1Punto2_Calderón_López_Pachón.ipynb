## Punto Dos.

Usando el diagrama de fase, realizar el análisis de los puntos críticos
para cada una de las siguientes ecuaciones autónomas.

a.  $y'=y(3-y)(y-2)$

```python
import numpy as np
import matplotlib.pyplot as plt # type: ignore

def f(x, y):
    return np.sin(y)

h = 0.5  # separacion de la malla
m = np.arange(-10, 11)
n = np.arange(-10, 11)
X, Y = np.meshgrid(m*h, n*h)

pendiente = f(X, Y)
norm = np.sqrt(1 + pendiente**2)
U = 1 / norm          # componente horizontal (unitaria)
V = pendiente / norm  # componente vertical

fig, ax = plt.subplots()
ax.quiver(X, Y, U, V, angles='xy', color='navy')  # flechas
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
plt.show()
```
