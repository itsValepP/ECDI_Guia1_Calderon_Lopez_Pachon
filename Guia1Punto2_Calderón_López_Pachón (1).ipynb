## Punto Dos.

Usando el diagrama de fase, realizar el análisis de los puntos críticos
para cada una de las siguientes ecuaciones autónomas.

### a.  $\small y'=y(3-y)(y-2)$

#### **Hallando los puntos críticos**

Iniciaremos por igualar $\dfrac{dy}{dx}$ a cero:

$$y(3-y)(y-2) = 0 \;\Rightarrow\; y = 0 \quad \text{o} \quad y = 2 \quad \text{o} \quad y = 3$$

Ahora analizaremos los signos en cada intervalo

| Intervalo        | $y$ | $3 - y$ | $y - 2$ | $dy/dx$ | Comportamiento |
|------------------|-----|--------|-----------|----------|----------------|
| $y<0$             |  $-$ |  $+$  | $-$  |  $+$               | crece        |
| $0<y<2$         |  $+$ |  $+$  | $-$ |  $-$               | decrece          |
| $2<y<3$           |  $+$ |  $+$  | $+$  |  $+$               | crece        |
| $y>3$             |  $+$ |  $-$   | $+$ |  $-$               | decrece        |

Por lo tanto:
* $P=0$ es un punto crítico atractor, es decir que es **estable**.
* $P=2$ es un punto crítico retractor , es decir que es **inestable**.
* $P=0$ es un punto crítico atractor, es decir que es **estable**.

#### **Graficando la solución**

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(3, 3))

# linea vertical
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-1, 1)
ax.set_ylim(-1.5, 4)

criticos_num = [0, 2, 3]
for pc in criticos_num:
    ax.plot(0, pc, 'o', color='red', markersize=8, zorder=3)
    ax.text(0.15, pc, f"P={pc}", va='center')

flechas = [
    (0, -1.2, 0, -0.35),   # y<0 crece
    (0, 1.6, 0, 0.4),      # 0<y<2 decrece
    (0, 2.15, 0, 2.85),    # 2<y<3 crece
    (0, 3.85, 0, 3.15),    # y>3 decrece
]
for x0, y0, dx, dy in flechas:
    ax.annotate('', xy=(0, dy), xytext=(0, y0),
                arrowprops=dict(arrowstyle='-|>', color='navy', lw=2))

ax.set_yticks(criticos_num)
ax.set_xticks([])
ax.set_ylabel("y")
ax.set_title("Diagrama de fase")
plt.tight_layout()
plt.show()
```

### b.  $y'=y^2-y^3$

Inicialmente vamos a reescribir la ecuación para que se nos facilite identificar los puntos críticos

$$y'=y^2(1-y)$$

#### **Hallando los puntos críticos**

Tras igualar  $\dfrac{dy}{dx}$ a cero, obtenemos:

$$y^2(1-y) = 0 \;\Rightarrow\; y = 0 \quad \text{o} \quad y = 1$$

Ahora analizaremos los signos en cada intervalo

| Intervalo        | $y^2$ | $1 - y$ | $dy/dx$ | Comportamiento |
|------------------|-----|--------|----------|----------------|
| $y<0$             |  $+$ |  $+$  |  $+$               | crece        |
| $0<y<1$         |  $+$ |  $+$  |  $+$               | crece          |
| $y>1$             |  $+$ |  $-$   |  $-$               | decrece        |

Por lo tanto:
* $P=0$ es un punto crítico a **semiestable**.
* $P=1$ es un punto crítico atractor , es decir que es **estable**.
#### **Graficando la solución**

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(3, 3))

# linea vertical
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-1, 1)
ax.set_ylim(-1, 2)

criticos_num = [0, 1]
for pc in criticos_num:
    ax.plot(0, pc, 'o', color='red', markersize=8, zorder=3)
    ax.text(0.15, pc, f"P={pc}", va='center')

flechas = [
    (0, -0.8, 0, -0.15),   # y<0 crece
    (0, 0.15, 0, 0.85),    # 0<y<1 crece
    (0, 1.8, 0, 1.15),     # y>1 decrece
]
for x0, y0, dx, dy in flechas:
    ax.annotate('', xy=(0, dy), xytext=(0, y0),
                arrowprops=dict(arrowstyle='-|>', color='navy', lw=2))

ax.set_yticks(criticos_num)
ax.set_xticks([])
ax.set_ylabel("y")
ax.set_title("Diagrama de fase")
plt.tight_layout()
plt.show()
```

### c.  $\small y'=(y+2)(10+3y-y^2)$

Iniciaremos por reescribir la ecuación para facilitar la identificación de los puntos críticos

$$(y+2)(10+3y-y^2) = (y+2)[-(y-5)(y+2)] = -(y+2)^2(y-5)$$

#### **Hallando los puntos críticos**

Al igualar  $\dfrac{dy}{dx}$ a cero, obtenemos:

$$-(y+2)^2(y-5) \;\Rightarrow\; y = -2 \quad \text{o} \quad y = 5$$

Ahora analizaremos los signos en cada intervalo

| Intervalo        | $-(y+2)^2$ | $y-5$ | $dy/dx$ | Comportamiento |
|------------------|-----|--------|----------|----------------|
| $y<-2$             |  $-$ |  $-$  |  $+$               | crece        |
| $-2<y<5$         |  $-$ |  $-$  |  $+$               | crece          |
| $y>5$             |  $-$ |  $+$   |  $-$               | decrece        |

Por lo tanto:
* $P=-2$ es un punto crítico a **semiestable**.
* $P=5$ es un punto crítico atractor , es decir que es **estable**.
#### **Graficando la solución**

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(3, 3))

# linea vertical
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-1, 1)
ax.set_ylim(-4, 7)

criticos_num = [-2, 5]
for pc in criticos_num:
    ax.plot(0, pc, 'o', color='red', markersize=8, zorder=3)
    ax.text(0.15, pc, f"P={pc}", va='center')

flechas = [
    (0, -3.7, 0, -2.5),   # y<-2 crece
    (0, -1.5, 0, 4.3),    # -2<y<5 crece
    (0, 6.7, 0, 5.5),     # y>5 decrece
]
for x0, y0, dx, dy in flechas:
    ax.annotate('', xy=(0, dy), xytext=(0, y0),
                arrowprops=dict(arrowstyle='-|>', color='navy', lw=2))

ax.set_yticks(criticos_num)
ax.set_xticks([])
ax.set_ylabel("y")
ax.set_title("Diagrama de fase")
plt.tight_layout()
plt.show()
```

### d.  $\small y'=y^5-4y^3-5y^2$

Iniciaremos por reescribir la ecuación con miras de identificar mejor los puntos críticos.

$$y^5-4y^3-5y^2 = y^2(y^3 -4y -5)$$

#### **Hallando los puntos críticos**

Al igualar  $\dfrac{dy}{dx}$ a cero, llegamos a lo siguiente:

$$y^2(y^3 -4y -5) \;\Rightarrow\; y = 0 \quad \text{o} \quad y = 2.457$$

Ahora analizaremos los signos en cada intervalo

| Intervalo        | $y^2$ | $(y^3 -4y -5)$ | $dy/dx$ | Comportamiento |
|------------------|-----|--------|----------|----------------|
| $y<0$             |  $+$ |  $-$  |  $-$               | decrece        |
| $0<y<2.457$         |  $+$ |  $-$  |  $-$               | decrece          |
| $y>2.457$             |  $+$ |  $+$   |  $+$               | crece        |

Por lo tanto:
* $P=0$ es un punto crítico **semiestable**.
* $P=2.457$ es un punto crítico retractor , es decir que es **inestable**.
#### **Graficando la solución**

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(3, 3))

# linea vertical
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-1, 1)
ax.set_ylim(-2, 4)

criticos_num = [0, 2.457]
for pc in criticos_num:
    ax.plot(0, pc, 'o', color='red', markersize=8, zorder=3)
    ax.text(0.15, pc, f"P={pc}", va='center')

flechas = [
    (0, -0.4, 0, -1.6),    # y<0 decrece
    (0, 2.15, 0, 0.35),    # 0<y<2.457 decrece
    (0, 2.75, 0, 3.7),     # y>2.457 crece
]
for x0, y0, dx, dy in flechas:
    ax.annotate('', xy=(0, dy), xytext=(0, y0),
                arrowprops=dict(arrowstyle='-|>', color='navy', lw=2))

ax.set_yticks(criticos_num)
ax.set_xticks([])
ax.set_ylabel("y")
ax.set_title("Diagrama de fase")
plt.tight_layout()
plt.show()
```
